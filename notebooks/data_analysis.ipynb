```python
import pandas as pd
```

```python
df = pd.read_csv("Braxton_data_jan_2024_dec16_2025.csv") 
df
```

```python
import matplotlib.pyplot as plt
```

```python
postal_units = (
    df.groupby("customer_postal_code")["unit_sold"]
      .sum()
      .sort_values(ascending=False)
)

# Plot top 10 to keep it readable
postal_units_top = postal_units.head(10)

plt.figure()
postal_units_top.plot(kind="bar")
plt.title("Units Sold per Postal Code (Top 10)")
plt.xlabel("Customer Postal Code")
plt.ylabel("Units Sold")
plt.xticks(rotation=45)
plt.show()
```

```python
city_units = (
    df.groupby("customer_city")["unit_sold"]
      .sum()
      .sort_values(ascending=False)
)

# Plot top 10 cities
city_units_top = city_units.head(10)

plt.figure()
city_units_top.plot(kind="bar")
plt.title("Units Sold per City (Top 10)")
plt.xlabel("Customer City")
plt.ylabel("Units Sold")
plt.xticks(rotation=45)
plt.show()
```
